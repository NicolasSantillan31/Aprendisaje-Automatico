```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Automobile.csv")
df.head()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["weight"], df["mpg"], color="red")
plt.xlabel("Peso")
plt.ylabel("MPG")
plt.title("Gráfico de peso y consumo")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["mpg"], bins=15, color="orange")
plt.title("Histograma del MPG")
plt.xlabel("MPG")
plt.ylabel("Cantidad de autos")
plt.title("Distribución del consumo de combustible")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.boxplot([df["cylinders"]])

plt.xticks([1], ["Cylinders"])

plt.title("Boxplot de Cylinders")
plt.grid(True)
plt.show()
```

```python
df["model_year"].value_counts().sort_index().plot(kind="bar", figsize=(10,5))

plt.title("Autos por año")
plt.xlabel("Año del modelo")
plt.ylabel("Cantidad de autos")
plt.grid(True)
plt.show()
```

## Conclusión

En este trabajo pude usar Matplotlib para graficar datos del archivo Automobile.csv.

Se pudo observar que los autos con mayor peso suelen consumir más combustible. También se vio la distribución del mpg y la cantidad de autos por año.

Estos gráficos ayudan a entender mejor la información y son útiles en aprendizaje automático.
